```python
import pandas as pd
import numpy as np

train_df = pd.read_csv("../data/processed/train.csv")
validation_df = pd.read_csv("../data/processed/validation.csv")
test_df = pd.read_csv("../data/processed/test.csv")

X_train = train_df.drop(columns=["phishing"])
y_train = train_df["phishing"]

X_val = validation_df.drop(columns=["phishing"])
y_val = validation_df["phishing"]

X_test = test_df.drop(columns=["phishing"])
y_test = test_df["phishing"]

print("Training shape:", X_train.shape)
print("Validation shape:", X_val.shape)
print("Test shape:", X_test.shape)
```

```python
from sklearn.feature_selection import SelectKBest, f_classif

selector = SelectKBest(score_func=f_classif, k="all")
selector.fit(X_train, y_train)

feature_scores = pd.DataFrame({
    "feature": X_train.columns,
    "f_score": selector.scores_,
    "p_value": selector.pvalues_
})

feature_scores = feature_scores.sort_values(
    "f_score",
    ascending=False
)

print(feature_scores.round(4))
```

```python
selected_features = feature_scores[
    feature_scores["p_value"] < 0.05
]["feature"].tolist()

print("Selected features:")
print(selected_features)

print("\nNumber of selected features:", len(selected_features))
```

```python
X_train_selected = train_df[selected_features]
X_val_selected = validation_df[selected_features]
X_test_selected = test_df[selected_features]

y_train = train_df["phishing"]
y_val = validation_df["phishing"]
y_test = test_df["phishing"]

print("Selected feature dataset shapes:")
print("Training:", X_train_selected.shape)
print("Validation:", X_val_selected.shape)
print("Test:", X_test_selected.shape)
```
